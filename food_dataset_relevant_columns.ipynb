```python
import pandas as pd
```

```python
food_data = pd.read_csv("food_dataset/en.openfoodfacts.org.products.tsv", sep='\t', low_memory=False)
food_data.head(5)
```

```python
food_data = food_data[food_data["countries_en"] == "United States"]
food_data["ingredients_text"].head(10)
```

```python
relevant_data = food_data[["product_name", "nutrition_grade_fr", "quantity", "ingredients_text", "additives_en", "energy_100g", "fat_100g", "saturated-fat_100g", "carbohydrates_100g", "sugars_100g", "proteins_100g", "fiber_100g", "salt_100g"]]
relevant_data.head(10)
```

```python
relevant_data.to_csv("food_dataset/relevant_data.csv", index = False)
```
